## 2. Create a client

Don't hardcode your key in a shared notebook — read it from the environment, or
use `getpass` to type it without it being saved in the notebook.

```python
import os
from getpass import getpass

from cspan import CSpanClient, AuthenticationError, ValidationError

# Prefer the CSPAN_API_KEY env var; fall back to a secure prompt.
if not os.environ.get("CSPAN_API_KEY"):
    os.environ["CSPAN_API_KEY"] = getpass("C-SPAN API key: ")

client = CSpanClient()   # reads CSPAN_API_KEY
client
```

## 3. Search spoken mentions

`GET /mentions` — find words or phrases spoken across C-SPAN programming.
Every method returns decoded JSON.

```python
mentions = client.mentions(
    "artificial intelligence",
    limit=2,
    mindate="2024-01-01",
)
mentions
```

## Choosing an output format

The API always returns JSON, but the client can convert it for you. Pick a
format per call with `format=`, or set `output_format=` on the client for every
call. Options: `"json"` (default), `"records"`, `"csv"`, `"dataframe"`.

```python
# Same query, different output formats
print(client.people(last="Pelosi", format="csv"))       # CSV string
records = client.people(last="Pelosi", format="records")  # list[dict]
df = client.people(last="Pelosi", format="dataframe")     # pandas DataFrame (needs pandas)
df
```

## 7. Pagination — built in

`iter_records` follows the `cursor` automatically and yields one row at a time.
Works with `bills`, `mentions`, `people`, and `programs_search`. Use `max_items`
to cap how many rows you pull.

```python
rows = list(client.iter_records(client.bills, query="infrastructure", max_items=50))
print(f"pulled {len(rows)} rows")
rows[:3]
```

## Error handling

Typed exceptions: `ValidationError` (bad input, raised before the request),
`AuthenticationError` (401/403), `NotFoundError` (404), `RateLimitError` (429),
and `APIError` (other errors). All inherit from `CSpanError`.

```python
from cspan import APIError, NotFoundError, RateLimitError

# ValidationError fires before any network call:
try:
    client.mentions("ai", date="2024-13-01")     # bad month
except ValidationError as exc:
    print("ValidationError:", exc)

# API-side errors map to typed exceptions carrying .status_code:
try:
    CSpanClient("bogus-key").people(last="Pelosi")
except AuthenticationError as exc:
    print("AuthenticationError:", exc, "| status:", exc.status_code)
```
